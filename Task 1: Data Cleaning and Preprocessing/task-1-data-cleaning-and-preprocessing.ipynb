```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("joniarroba/noshowappointments")

print("Path to dataset files:", path)
```

```python
import pandas as pd

m_data = pd.read_csv('/kaggle/input/datasets/joniarroba/noshowappointments/KaggleV2-May-2016.csv')

m_data.head(5)
```

```python
# Describing the dataset
m_data.info()
```

```python
# Identifying & handling the missing values

m_data.isnull().sum()
```

```python
m_data = m_data.dropna()

m_data.isnull().sum()
```

```python
# Removing the duplicates

m_data.drop_duplicates()

m_data.tail(5)
```

```python
# Standardizing text values
# Renaming the columns

m_data.columns = (
    m_data.columns.str.strip()
    .str.lower()
    .str.replace('-','_')
)

m_data.columns
```

```python
# Formatting the dates

m_data['scheduledday'] = pd.to_datetime(m_data['scheduledday'])

m_data['appointmentday'] = pd.to_datetime(m_data['appointmentday'])
```

```python
# Fixing the datatypes

m_data['age'] = m_data['age'].astype(int)
m_data['patientid'] = m_data['patientid'].astype(int)

m_data.info()
```

```python
m_data['gender'] = m_data['gender'].replace({
    'M' : 'Male',
    'F' : 'Female'
})

m_data['gender'].unique()
```

```python
# Checking the values

m_data[m_data['age'] < 0]
```

```python
m_data = m_data[m_data['age']>=0]

m_data[m_data['age'] < 0]
```

```python
# Cheking the values 

m_data['no_show'].unique()
```

```python
# Saving the cleaned dataset

m_data.to_csv('/kaggle/working/cleaned_medical_appointments.csv', index=False)
```
